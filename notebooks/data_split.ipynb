```python
from pathlib import Path
```

```python
dir = '../data/tiny_shakespeare.txt'
path = Path(dir)
```

```python
text = path.read_text(encoding='utf-8')
```

```python
len(text)
```

```python
total_len = len(text)
ckpt1 = int(total_len*0.9) + 3
ckpt2 = int(total_len*0.95) + 57

train = text[:ckpt1]
val = text[ckpt1:ckpt2]
test = text[ckpt2:]

print(train[-10:-1])
print()
print(val[:10])
print('\n\n')
print(val[-10:-1])
print()
print(test[:10])
```

```python
len(train)
```

```python
len(test)
```

```python
ls = {
    'train': train,
    'val': val,
    'test': test
    }

for name, text_split in ls.items():
    path = Path(f'../data/tiny_shakespeare_{name}.txt')
    path.write_text(text_split)
```
