```python
fulltext = Path('../data/tiny_shakespeare.txt').read_text()
```

```python
len(fulltext)
```

```python
tokenizer = Tokenizer(BPE(unk_token='[UNK]'))
tokenizer.pre_tokenizer = ByteLevel()
```

```python
trainer = BpeTrainer(
    vocab_size=16000,
    special_tokens=["[UNK]", "[CLS]", "[SEP]", "[PAD]", "[MASK]"]
)
'''
토큰,의미,역할
[UNK],Unknown,사전에 없는 문자가 올 때 처리 (ByteLevel 덕분에 실제론 거의 안 쓰임)
[CLS],Classification,문장의 시작을 알림 (BERT 계열에서 주로 사용)
[SEP],Separator,문장 사이의 구분자
[PAD],Padding,배치의 길이를 맞추기 위한 빈 칸 채우기 (매우 중요)
[MASK],Masking,특정 단어를 가리고 맞추는 학습용
'''

tokenizer.train_from_iterator([fulltext], trainer=trainer)

tokenizer.save("../data/shakespeare_bpe.json")
```
