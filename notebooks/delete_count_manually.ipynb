```python
txt_to_drop = r"\[표\]|지선-표|후보-표|인사 종합|인사종합|△|정보공개|정보 공개|php|json"
for i in range(1, 33):
    dir = f'/home/hwajoong/projects/ido/data/korean_processed/token_count_added/part-0000{i:02d}.parquet'
    df = pd.read_parquet(dir)
    before = len(df)
    df.drop(
        index=df[df["content"].str.contains(txt_to_drop, na=False, case=False)].index,
        inplace=True
    )
    after = len(df)
    print(f'parquet-0000{i:02d} : {before} - {after} = {before - after}')
    df.to_parquet(dir, index=False)
    del df
```
