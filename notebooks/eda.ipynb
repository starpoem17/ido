```python
num = 1
root = '../data/korean_processed/_staging/novel24/val/part-'
while True:
    try:
        df = pd.read_parquet(f'{root}{num:06d}.parquet')
    except:
        print('file done')
        break
    ori_len = len(df)
    df.drop_duplicates(subset=['content'], keep='first', inplace=True)
    after_len = len(df)
    df.to_parquet(f'{root}{num:06d}.parquet', index=False)
    diff = ori_len - after_len
    if diff != 0:
        print(diff, num)
    num += 1
```
